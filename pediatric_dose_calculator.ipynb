```python
print("--(SAMPLE1 : DRUG DOSAGE CALCULATOR)--")
print("----By ABHISHEK VERMA---")
Patient_Name = input("Insert Patient Name (abc): ")
Age = int(input("Insert Patient Age (in years): "))
Adult_dose = float(input("Adult Dose(in mg): "))

if Age < 4:
    child_dose = (Age / Age + 12) * Adult_dose
    rule_used = "Young's Rule"
else:
    child_dose = (Age / 20) * Adult_dose
    rule_used = "Dilling's Rule"

print(f"--- Overall Prescription summary ---")
print(f"Patient Name: {Patient_Name}")
print(f"formula applies: {rule_used}")
print("Recommended Pediatric Dose:", round(child_dose, 2), "mg")
print(f"--THANKS--")
```
