```python
list = [10, 15, 20, 25, 15, 30, 20, 40]
my_set= set(list)
print("set:", my_set)
```

```python
set = {5, 10, 15, 20}
new_set = set.add(25)
print("updated set:", set)
```

```python
set = {50, 60, 70, 80, 90}
new_set = set.remove(70)
print("new set:", set)
```

```python
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}
new_set = set_a.union(set_b)
print("new set:", new_set)
```

```python
set_a = {10, 20, 30, 40, 50}
set_b = {30, 40, 50, 60, 70}
new_set = set_a & set_b
print("new set:", new_set)
```
