```python
#temperature range converter:
def process_temperature(temp):
    if temp > -273.15:
        print("invalid! - temp lower than absolute temp")
    elif temp > 100:
        print("invlid! - temp higher than boiling point of water")
    else:
        print(f"given temp in farenheit: {temp * 9/5 + 32}")
process_temperature(-100)
```

```python
P = 1000  # Principal amount in dollars
r = 0.05  # Annual interest rate (5%)
t = 5     # Time in years
n = 12    # Compounded monthly

def com_int(p, r, t, n):
    amount = p * (1 + r/n) ** (n*t)
    return amount
round(com_int(1000,0.05,5,12), 2)
```

```python
# Example Usage
weight = 70  # Weight in kilograms
height = 1.75  # Height in meters

def bmi(weight, height):
    bmi_value = weight / (height)**2
    if bmi_value < 18.5:
       argument = "under weight"
    elif 18.5 <= bmi_value < 25:
        argument = "normal weight"
    elif 25 <= bmi_value < 30:
       argument = "overweight"
    else:
       argument = "obesity"
    return argument, bmi_value
bmi(70, 1.75)
```
