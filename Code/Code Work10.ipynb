```python
class library:
    def __init__(self,name,author):
        self.name = name
        self.author = author
        self.is_available = True
    def borrow(self):
        if self.is_available:
            self.is_available = False
            return f"'{self.name}' by '{self.author}' is borrowed by you."
        else:
            return f"'{self.name}' by '{self.author}' is already borrowed."
    def return_book(self):
        if not self.is_available:
            self.is_available = True
            return f"'{self.name}' by '{self.author}' is returned by you."
        else:
             return f"'{self.name}' by '{self.author}' is already available in library."
    def check_availability(self):
        return f"{self.name} is {'available' if self.is_available else 'not available'}"
       
```

```python
book = library("Think straight", "Darius Forux")
```

```python
print(book.check_availability())
```

```python
print(book.borrow())
```

```python
print(book.check_availability())
```
